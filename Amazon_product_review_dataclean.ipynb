**Data cleaning**

**1. product_id**

step: remove "B0" from all value in product_id column

```python
df['product_id'] = df['product_id'].str.replace('B0','',regex=True)
display(df.head())
```

**Remove extra spaces**

```python
df["product_id"] = df["product_id"].str.strip()
```

```python
df["product_id"].duplicated().sum()
```

**2.Product Name**

**Fill missing values**

```python
df["product_name"] = df["product_name"].fillna("Unknown")
```

**Convert to Proper Case**

```python
df["product_name"] = df["product_name"].str.title()
```

**Remove extra spaces**

```python
df["product_id"] = df["product_id"].str.strip()
```

**Category**

Check missing values

```python
df["category"].isnull().sum()
```

Fill missing values

```python
df["category"] = df["category"].fillna("Unknown")
```

Remove extra spaces

```python
df["category"] = df["category"].str.strip()
```

**4.Discounted Price**

Remove ₹

```python
df["Discounted_Price (INR)"] = df["Discounted_Price (INR)"].str.replace("₹","",regex=False)
```

Remove Rs.

```python
df["Discounted_Price (INR)"] = df["Discounted_Price (INR)"].str.replace("Rs.","",regex=False)
```

Remove commas

```python
df["Discounted_Price (INR)"] = df["Discounted_Price (INR)"].str.replace(",","",regex=False)
```

Convert to Float

```python
df["Discounted_Price (INR)"] = pd.to_numeric(df["Discounted_Price (INR)"],errors="coerce")
```

Replace values less than or equal to 0

```python
df.loc[df["Discounted_Price (INR)"]<=0,"Discounted_Price (INR)"]=np.nan
```

Fill missing values

```python
df["Discounted_Price (INR)"]=df["Discounted_Price (INR)"].fillna(df["Discounted_Price (INR)"].median())
```

Show the table

```python
display(df.head())
```
